```python
import pandas as pd
import numpy as np

# 1.1 用字典创建DataFrame
data_dict = {
    'name': ['Tom', 'Jerry', 'Mike'],
    'score': [85, 92, 78],
    'gender': ['男', '女', '男']
}
df = pd.DataFrame(data_dict)
print("原始DataFrame:")
print(df)
print()

# 查看操作
print("前1行:")
print(df.head(1))
print()

print(f"形状: {df.shape}")
print()

print(f"列名: {df.columns.tolist()}")
print()

print(f"数据类型:\n{df.dtypes}")
print()

# 1.2 用列表创建DataFrame
data = [[1, "A"], [2, "B"], [3, "C"]]
df_list = pd.DataFrame(data, columns=['id', 'tag'])
print("列表创建的DataFrame:")
print(df_list)
print()
```

```python
# 选取score和gender两列
selected_cols = df[['score', 'gender']]
print("选取score和gender两列:")
print(selected_cols)
print()

# 选取索引0、2的行
selected_rows = df.iloc[[0, 2]]
print("选取索引0、2的行:")
print(selected_rows)
print()

# 筛选score>80的行
filtered = df[df['score'] > 80]
print("筛选score>80的行:")
print(filtered)
print()

# 新增一列grade
df['grade'] = ['优秀' if s >= 90 else '良好' for s in df['score']]
print("新增grade列:")
print(df)
print()

# 删除gender列
df_drop = df.drop('gender', axis=1)
print("删除gender列:")
print(df_drop)
print()
```

```python
# 创建含缺失值的DataFrame
missing_data = {"x": [1, None, 3], "y": [None, 5, 6]}
df_missing = pd.DataFrame(missing_data)
print("含缺失值的DataFrame:")
print(df_missing)
print()

# 检测缺失值
print("缺失值检测:")
print(df_missing.isnull())
print()

# 统计每列缺失数量
print("每列缺失数量:")
print(df_missing.isnull().sum())
print()

# 用0填充缺失值
df_filled = df_missing.fillna(0)
print("用0填充缺失值:")
print(df_filled)
print()

# 删除含缺失值的行
df_dropped = df_missing.dropna()
print("删除含缺失值的行:")
print(df_dropped)
print()
```

```python
# 重新创建原始df，因为之前修改过
df = pd.DataFrame({
    'name': ['Tom', 'Jerry', 'Mike'],
    'score': [85, 92, 78],
    'gender': ['男', '女', '男']
})

# 按gender分组，求score的均值和最大值
grouped = df.groupby('gender')['score'].agg(['mean', 'max'])
print("按gender分组聚合(mean, max):")
print(grouped)
print()

# 按gender分组，统计每组人数
group_count = df.groupby('gender').size()
print("按gender分组统计人数:")
print(group_count)
print()
```

```python
# 按score升序排序
df_sorted = df.sort_values('score')
print("按score升序排序:")
print(df_sorted)
print()

# 添加重复数据
df_dup = df.copy()
df_dup = df_dup._append({'name': 'Tom', 'score': 85, 'gender': '男'}, ignore_index=True)
print("添加重复行后:")
print(df_dup)
print()

# 去重
df_deduplicated = df_dup.drop_duplicates()
print("去重后:")
print(df_deduplicated)
print()
```

```python
# 横向合并
df1 = pd.DataFrame({"id": [1, 2], "name": ["X", "Y"]})
df2 = pd.DataFrame({"id": [1, 2], "age": [20, 30]})

merged = pd.merge(df1, df2, on='id')
print("横向合并:")
print(merged)
print()

# 纵向拼接
df_a = pd.DataFrame({"val": [1, 2]})
df_b = pd.DataFrame({"val": [3, 4]})

concatenated = pd.concat([df_a, df_b], ignore_index=True)
print("纵向拼接:")
print(concatenated)
print()
```

```python
# 创建时间列
df_time = pd.DataFrame({
    "date_str": ["2025-01-01", "2025-01-02", "2025-01-03"]
})

# 转为datetime格式
df_time['date'] = pd.to_datetime(df_time['date_str'])
print("时间处理:")
print(df_time)
print()

# 提取月份和星期
df_time['month'] = df_time['date'].dt.month
df_time['weekday'] = df_time['date'].dt.day_name()
print("提取月份和星期:")
print(df_time[['date', 'month', 'weekday']])
print()
```

```python
# 重新创建原始df
df = pd.DataFrame({
    'name': ['Tom', 'Jerry', 'Mike'],
    'score': [85, 92, 78],
    'gender': ['男', '女', '男']
})

# 对score列做标准化
score_mean = df['score'].mean()
score_std = df['score'].std()
df['score_standardized'] = (df['score'] - score_mean) / score_std
print("score标准化后:")
print(df[['name', 'score', 'score_standardized']])
print()

# 透视表
pivot_table = df.pivot_table(index='gender', values='score', aggfunc='median')
print("透视表（按gender分组，score中位数）:")
print(pivot_table)
```
