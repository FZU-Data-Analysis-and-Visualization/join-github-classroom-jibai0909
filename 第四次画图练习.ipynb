```python
import matplotlib.pyplot as plt
import numpy as np




years = ['2012年', '2013年', '2014年', '2015年', '2016年', '2017年', 
         '2018年', '2019年', '2020年', '2021年']

gdp_data = [53.8, 59.3, 64.4, 68.9, 74.6, 83.2, 91.9, 98.7, 101.6, 114.4]

growth_rate = [10.40, 7.80, 7.40, 7.00, 6.80, 6.90, 6.70, 6.00, 2.30, 8.10]

plt.rcParams['font.sans-serif'] = ['SimHei'] 
plt.rcParams['axes.unicode_minus'] = False  

fig, ax1 = plt.subplots(figsize=(10, 6)) 

#绘制柱状图

bars = ax1.bar(years, gdp_data, color='#1f77b4', width=0.6, label='GDP(万亿)')

ax1.set_ylabel('GDP(万亿)', fontsize=16)
ax1.set_ylim(0, 125)  
ax1.set_yticks(np.arange(0, 126, 25))  

for bar in bars:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2, height + 1,  
             f'{height}',                                   
             ha='center', va='bottom', fontsize=12)
ax1.set_yticklabels([f'{i:.1f}' for i in np.arange(0, 126, 25)])

#绘制折线图 
ax2 = ax1.twinx()

line = ax2.plot(years, growth_rate, color="#51b8cf", marker='o', linestyle='-', linewidth=2, label='增长率')

ax2.set_ylabel('增长率', fontsize=16)
ax2.set_ylim(0, 20)  
ax2.set_yticks(np.arange(0, 21, 4))  
ax2.set_yticklabels([f'{i:.1f}%' for i in range(0, 21, 4)])

for i, rate in enumerate(growth_rate):
    ax2.text(i, rate + 0.3, f'{rate}%', ha='center', va='bottom', fontsize=12)

ax1.tick_params(axis='x', labelsize=14)

#美化图表
ax1.spines['top'].set_visible(False)
ax1.spines['left'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
ax2.spines['left'].set_visible(False)

plt.title('组合图', fontsize=20, pad=20)  

ax1.grid(axis='y', linestyle='--', alpha=0.4, color='gray')
ax1.set_axisbelow(True)

lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
ax1.legend(lines_1 + lines_2, labels_1 + labels_2, loc='upper left')

plt.show()
```
